```python
labels = ["BENIGN", "DoS", " benign", "PortScan", "Dos", "BruteForce"]

is_attack = []

label=[item.strip().lower() for item in labels]

for label in label:
    print(label)
    if label in ["portscan","dos","bruteforce"]:
        is_attack.append(1)
    else:
        is_attack.append(0)

print(is_attack)
```
